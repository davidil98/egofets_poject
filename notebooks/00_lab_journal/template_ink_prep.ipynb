# OSC Ink Preparation Log

Record each batch of semiconductor ink (diFT:PS in chlorobenzene).

**Target:** 2% weight/volume SC + PS in CB, 4:1 ratio SC:PS  
**Rule of thumb:** 22.6 mg reagent per mL CB

```python
import sys, os
from datetime import datetime
PROJECT_ROOT = os.path.abspath(os.path.join(os.getcwd(), '../..'))
sys.path.insert(0, PROJECT_ROOT)

from src.database import init_db, add_ink_batch, list_ink_batches
import pandas as pd

init_db()

print('Existing ink batches:')
inks = list_ink_batches()
if inks:
    df = pd.DataFrame(inks)
    display(df[['name', 'sc_weight_mg', 'ps_weight_mg', 'cb_volume_ml', 'ratio_osc_ps', 'prepared_at']])
else:
    print('  (none)')
```

## 1. Weighing

Weigh the SC (diFT-TES-ADT) and PS (10k g/mol) on the balance.

```python
SC_WEIGHT_MG = 4.5   # Actual weighed mass
PS_WEIGHT_MG = 1.1   # Actual weighed mass
CB_VOLUME_ML = 0.25  # Chlorobenzene volume

# Verify concentration (should be ~2% = 22.6 mg/mL)
sc_conc = SC_WEIGHT_MG / CB_VOLUME_ML
ps_conc = PS_WEIGHT_MG / CB_VOLUME_ML
print(f'SC concentration: {sc_conc:.1f} mg/mL (target: 22.6)')
print(f'PS concentration: {ps_conc:.1f} mg/mL (target: 22.6)')
print(f'Actual SC %: {sc_conc / 22.6 * 2:.2f}%')
print(f'Actual PS %: {ps_conc / 22.6 * 2:.2f}%')
print(f'Total volume: {CB_VOLUME_ML * 1000:.0f} µL')
print(f'Estimated devices: ~{int(CB_VOLUME_ML * 1000 / 35)} (at 35 µL/device)')
```

## 2. Chlorobenzene Extraction

Extract CB from bottle under Ar atmosphere:
1. Inflate balloon with Ar, insert small needle as septum
2. Extract CB with syringe + large needle (DO NOT tip bottle)
3. Transfer to amber vial

## 3. Dissolution

- Dissolve SC and PS separately in CB (amber vials covered with Al foil)
- Place on BAMS hotplate at **105°C for ≥ 60 min**
- Mix SC:PS at **4:1 ratio**
- Leave **overnight at 60°C** for further use

```python
INK_NAME = datetime.now().strftime('INK-%y%m%d')
SC_MATERIAL = 'diFT-TES-ADT'
PS_MW = 10000
RATIO = '4:1'
NOTES = ''

ink_id = add_ink_batch(
    name=INK_NAME,
    sc_weight_mg=SC_WEIGHT_MG,
    ps_weight_mg=PS_WEIGHT_MG,
    cb_volume_ml=CB_VOLUME_ML,
    sc_material=SC_MATERIAL,
    ps_mw=PS_MW,
    ratio=RATIO,
    notes=NOTES,
)

print(f'Ink batch **{INK_NAME}** registered (id={ink_id}).')
print(f'Usable after: ~24h (overnight at 60°C)')
print(f'\nSC: {SC_WEIGHT_MG} mg {SC_MATERIAL}')
print(f'PS: {PS_WEIGHT_MG} mg (MW={PS_MW})')
print(f'CB: {CB_VOLUME_ML} mL')
print(f'Ratio: {RATIO}')
```

---
### Calculation Reference

For 2% weight in CB:
- **22.6 mg reagent per mL CB**
- Per device: ~30-40 µL
- For N devices, estimate volume V = N × 35 µL
- Weigh reagent ≈ 22.6 × V(mL) mg, then recalculate with actual weight

Example: 10 devices → 350 µL → 22.6 × 0.35 ≈ 7.9 mg SC  
At 4:1, PS needed = SC / 4 ≈ 2.0 mg
